# DNN 用于语音识别：PyTorch 交互式课堂 Notebook

本 notebook 使用 `asr-teaching-assistant` 的课堂结构组织：每完成一个关键步骤，就通过 checkpoint 检查学生是否理解。学生答错时，系统会给出提示和补充讲解，并记录学习表现。

本节目标：理解 DNN 声学模型在语音识别中的基本位置：声学特征输入 -> DNN 分类模型 -> logits -> 损失函数训练 -> 预测与评价 -> 解码意识。

```python
import sys
import random
import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
checkpoint_records = []

print('Python:', sys.version.split()[0])
print('PyTorch:', torch.__version__)
print('Device:', device)
```

```python
def ask_until_correct(title, question, answer_set, hint, knowledge, max_attempts=3):
    """课堂 checkpoint：答对继续；答错则补讲，并记录学习诊断。"""
    print('\n【知识点检查】' + title)
    answer_set = {str(x).strip().lower() for x in answer_set}
    attempts = []
    for attempt in range(1, max_attempts + 1):
        user_answer = input(question + '\n请输入答案：').strip().lower()
        attempts.append(user_answer)
        if user_answer in answer_set:
            status = '已掌握' if attempt == 1 else '部分掌握'
            print('回答正确，可以进入下一步。')
            checkpoint_records.append({
                'title': title,
                'question': question,
                'answers': attempts,
                'status': status,
                'feedback': '能抓住本知识点的关键含义。'
            })
            return status
        print('\n回答还不准确。')
        print('提示：' + hint)
        print('补充知识：' + knowledge + '\n')

    print('这个知识点暂时标记为“需教师介入”，课堂继续推进，课后再补学。')
    checkpoint_records.append({
        'title': title,
        'question': question,
        'answers': attempts,
        'status': '需教师介入',
        'feedback': knowledge
    })
    return '需教师介入'
```

## Step 1 任务背景：DNN 声学模型学什么？

真实 ASR 系统会从音频中提取 MFCC、FBank 或谱图等声学特征，再由声学模型估计音素、字符、子词或词类别的概率。本节先用一个小型合成数据集模拟“声学特征 -> 数字类别”的过程，重点理解 DNN 声学模型的训练逻辑。

```python
ask_until_correct(
    title='DNN 声学模型的输入',
    question='我刚才为什么说 DNN 声学模型通常输入分帧后的声学特征，而不是直接输入整段原始 wav？回答：声学特征 / 原始wav',
    answer_set={'声学特征', '特征', '分帧后的声学特征'},
    hint='想一想 MFCC、FBank、谱图为什么常出现在语音识别流程里。',
    knowledge='语音在很短时间内近似平稳，所以常先分帧，再提取 MFCC、FBank 或谱图等声学特征。DNN 学的是这些特征和类别之间的映射，而不是直接理解整段 wav 的全部变化。'
)
```

## Step 2 构造简化声学特征数据

这里构造四类英文数字 `zero/one/two/three` 的二维特征分布。二维只是为了课堂演示；真实语音特征通常有更高维度，并且随时间形成序列。

```python
labels = np.array(['zero', 'one', 'two', 'three'])
centers = np.array([
    [0.0, 0.0],
    [2.5, 0.5],
    [0.5, 2.5],
    [2.6, 2.7],
], dtype=np.float32)

X_parts, y_parts = [], []
for class_id, center in enumerate(centers):
    features = center + np.random.normal(scale=0.55, size=(120, 2)).astype(np.float32)
    X_parts.append(features)
    y_parts.append(np.full(120, class_id, dtype=np.int64))

X = np.vstack(X_parts).astype(np.float32)
y = np.concatenate(y_parts).astype(np.int64)

print('样本数:', X.shape[0])
print('特征维度:', X.shape[1])
print('类别:', labels.tolist())
```

```python
ask_until_correct(
    title='特征维度与类别数',
    question='我刚才构造的 X 有 2 个特征维度、4 个类别。DNN 输入层维度应该对应哪个？回答：特征维度 / 类别数',
    answer_set={'特征维度', '特征', '2'},
    hint='输入层接收的是每个样本本身的表示。',
    knowledge='DNN 输入层维度对应每个样本的特征维度；输出层维度才通常对应类别数。本例输入维度是 2，输出维度是 4。'
)
```

## Step 3 划分数据与构造 DataLoader

训练集用于学习参数，测试集用于模拟模型没见过的新语音。标准化只在训练集上 `fit`，再应用到测试集，避免测试集信息提前泄漏。

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=SEED, stratify=y
)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train).astype(np.float32)
X_test_scaled = scaler.transform(X_test).astype(np.float32)

train_dataset = TensorDataset(torch.tensor(X_train_scaled), torch.tensor(y_train))
test_tensor = torch.tensor(X_test_scaled).to(device)
train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)

print('训练集:', X_train_scaled.shape)
print('测试集:', X_test_scaled.shape)
print('batch 数:', len(train_loader))
```

```python
ask_until_correct(
    title='训练集与测试集',
    question='我刚才为什么要留出测试集，而不是只看训练集准确率？回答：泛化 / 训练更快',
    answer_set={'泛化', '评价泛化', '测试泛化', '防止过拟合'},
    hint='训练集表现好，不等于新语音也识别得好。',
    knowledge='测试集用于估计模型对未见样本的表现。只看训练集准确率，容易把过拟合误认为模型能力强。语音识别最终关心的是新说话人、新环境、新语音上的表现。'
)
```

## Step 4 定义 PyTorch DNN 声学模型

这个模型是一个前馈 DNN。它接收二维声学特征，输出四个数字类别的 logits。logits 不是概率，而是 softmax 之前的原始分类分数。

```python
class AcousticDNN(nn.Module):
    def __init__(self, input_dim=2, hidden_dim=32, output_dim=4):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, hidden_dim // 2),
            nn.ReLU(),
            nn.Linear(hidden_dim // 2, output_dim)
        )

    def forward(self, x):
        return self.net(x)


model = AcousticDNN(input_dim=2, hidden_dim=32, output_dim=4).to(device)
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)

print(model)
```

```python
ask_until_correct(
    title='输出层与类别数',
    question='我刚才把 DNN 输出层设成 4 个单元，它对应本任务的什么？回答：类别数 / 特征维度',
    answer_set={'类别数', '类别', '4'},
    hint='每个输出单元对应一个候选类别的分数。',
    knowledge='分类模型的输出层单元数通常等于类别数。本例要识别 zero、one、two、three 四个数字，所以输出层是 4 个单元。'
)
```

## Step 5 训练循环：logits、loss 与反向传播

`CrossEntropyLoss` 直接接收 logits 和整数类别标签。训练时要先清空旧梯度，再反向传播并更新参数。

```python
ask_until_correct(
    title='CrossEntropyLoss 与 logits',
    question='我刚才准备用 CrossEntropyLoss 训练模型。它通常接收 logits，还是已经 softmax 后的概率？回答：logits / 概率',
    answer_set={'logits', 'logit'},
    hint='PyTorch 的 CrossEntropyLoss 内部已经包含 log-softmax 相关计算。',
    knowledge='在 PyTorch 中，CrossEntropyLoss 通常直接接收未经过 softmax 的 logits。如果先手动 softmax，可能影响数值稳定性，也和该损失函数的预期输入不一致。'
)
```

```python
epochs = 80
loss_history = []

for epoch in range(1, epochs + 1):
    model.train()
    total_loss = 0.0
    for batch_x, batch_y in train_loader:
        batch_x = batch_x.to(device)
        batch_y = batch_y.to(device)

        logits = model(batch_x)
        loss = criterion(logits, batch_y)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        total_loss += loss.item() * batch_x.size(0)

    avg_loss = total_loss / len(train_dataset)
    loss_history.append(avg_loss)
    if epoch == 1 or epoch % 20 == 0:
        print(f'Epoch {epoch:03d}, loss = {avg_loss:.4f}')
```

```python
ask_until_correct(
    title='梯度清零',
    question='我刚才每个 batch 都执行 optimizer.zero_grad()，主要是为了防止什么？回答：梯度累积 / 数据泄漏',
    answer_set={'梯度累积', '防止梯度累积', '累积'},
    hint='PyTorch 默认会把梯度加到已有梯度上。',
    knowledge='如果不清零，当前 batch 的梯度会和上一个 batch 的梯度叠加，导致参数更新不符合当前 mini-batch 的训练意图。'
)
```

## Step 6 评估与简单解码

评估时使用 `model.eval()` 和 `torch.no_grad()`，关闭训练相关行为并避免记录梯度。这里用 `argmax` 把 logits 转成类别 id，再映射回数字标签。

```python
model.eval()
with torch.no_grad():
    test_logits = model(test_tensor)
    pred_ids = torch.argmax(test_logits, dim=1).cpu().numpy()

acc = accuracy_score(y_test, pred_ids)
print('测试集准确率:', round(acc, 4))
print('\n混淆矩阵：')
print(confusion_matrix(y_test, pred_ids))
print('\n分类报告：')
print(classification_report(y_test, pred_ids, target_names=labels))

decoded_words = labels[pred_ids[:10]]
print('前 10 个预测标签:', decoded_words.tolist())
```

```python
ask_until_correct(
    title='评估模式与解码',
    question='我刚才评估时使用 model.eval() 和 torch.no_grad()，主要是为了训练参数，还是为了稳定评估并关闭梯度记录？回答：稳定评估 / 训练参数',
    answer_set={'稳定评估', '关闭梯度记录', '稳定评估并关闭梯度记录'},
    hint='评估阶段不需要反向传播。',
    knowledge='model.eval() 会让 dropout、batchnorm 等层进入评估行为；torch.no_grad() 不记录梯度，节省内存并避免误以为评估也在训练模型。'
)
```

## Step 7 错误分析与课堂迁移

本例是单词类别分类，所以可以用 accuracy 和混淆矩阵。完整连续语音识别输出的是字符、词或子词序列，常用 CER/WER 评价插入、删除、替换等序列错误。

```python
wrong_mask = pred_ids != y_test
wrong_indices = np.where(wrong_mask)[0]
print('错误样本数量:', len(wrong_indices))
for idx in wrong_indices[:5]:
    print('样本', int(idx), '真实:', labels[y_test[idx]], '预测:', labels[pred_ids[idx]], '特征:', X_test_scaled[idx])
```

```python
ask_until_correct(
    title='从分类到完整 ASR',
    question='如果从本例的数字分类扩展到完整句子识别，只看 accuracy 够不够？回答：不够 / 够',
    answer_set={'不够', '不', '不够用'},
    hint='完整 ASR 输出的是变长文字序列，不只是单个类别。',
    knowledge='完整 ASR 常用 CER/WER，因为模型可能出现替换、插入、删除等序列错误。普通 accuracy 适合固定类别分类，但不能充分描述一句话里错了几个字或词。'
)
```

## Step 8 自动生成学习反馈报告

运行下面的 cell，可以根据本 notebook 中的 checkpoint_records 生成课堂学习反馈。教师可将输出保存为学习过程记录。

```python
from collections import Counter

def build_learning_report(records):
    counts = Counter(record['status'] for record in records)
    lines = []
    lines.append('# 学生学习反馈报告：DNN 用于语音识别')
    lines.append('')
    lines.append('## 1. 任务摘要')
    lines.append('- 课程章节：DNN 声学模型用于语音识别')
    lines.append('- 技术栈：PyTorch、NumPy、scikit-learn 辅助划分与评价')
    lines.append('- 实验目标：理解声学特征输入、DNN 输出层、logits、训练循环、评估与解码意识')
    lines.append(f'- 测试集准确率：{acc:.4f}')
    lines.append('')
    lines.append('## 2. Checkpoint 统计')
    for status in ['已掌握', '部分掌握', '需教师介入']:
        lines.append(f'- {status}：{counts.get(status, 0)}')
    lines.append('')
    lines.append('## 3. 逐项诊断')
    for i, record in enumerate(records, start=1):
        answer_text = '；'.join(record['answers']) if record['answers'] else '未作答'
        lines.append(f"{i}. {record['title']}：{record['status']}。学生回答：{answer_text}。反馈：{record['feedback']}")
    lines.append('')
    lines.append('## 4. 后续建议')
    lines.append('- 若“声学特征输入”未掌握，回看分帧、MFCC/FBank、谱图的区别。')
    lines.append('- 若“logits 与 CrossEntropyLoss”未掌握，重点复习 softmax 前后的含义。')
    lines.append('- 若“评估与解码”未掌握，进一步学习 CER/WER 与 CTC 解码。')
    return '\n'.join(lines)

report = build_learning_report(checkpoint_records)
print(report)
```
