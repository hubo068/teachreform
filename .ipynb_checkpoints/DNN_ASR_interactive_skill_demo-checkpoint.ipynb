# DNN 用于语音识别：交互式课堂训练示例

本 notebook 演示一种“AI 大模型 skills + Python notebook”的课堂组织方式：学生在关键知识点处必须回答问题，答对后继续实验；答错时系统补充知识点，并要求再次作答。

教学目标：理解 DNN 语音识别的基本流程：音频信号 → 分帧/特征提取 → DNN 声学模型 → 类别/音素后验概率 → 解码与评价。

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

np.random.seed(42)
print('环境准备完成：本示例使用 sklearn 的 MLPClassifier 模拟 DNN 声学模型。')
```

```python
def ask_until_correct(title, question, answer_set, hint, knowledge):
    """课堂交互：学生答对才继续；答错则补充知识点。"""
    print('\n【知识点检查】' + title)
    while True:
        user_answer = input(question + '\n请输入答案：').strip().lower()
        if user_answer in answer_set:
            print('回答正确，可以进入下一步。')
            break
        print('\n回答还不准确。')
        print('提示：' + hint)
        print('补充知识：' + knowledge + '\n')
```

```python
ask_until_correct(
    title='语音识别中的 DNN 输入',
    question='DNN 声学模型通常直接输入原始整段 wav，还是输入分帧后的声学特征？回答：原始wav / 声学特征',
    answer_set={'声学特征', '特征', '分帧后的声学特征'},
    hint='想一想 MFCC、FBank、谱图这类表示的作用。',
    knowledge='传统 DNN-HMM 或早期端到端前的 DNN 声学模型，常以分帧后的 MFCC、FBank 或谱图特征作为输入，用神经网络估计每一帧属于某个音素、状态或字符类别的概率。'
)
```

## 1. 构造一个简化的“语音特征”数据集

真实语音识别会从音频中提取 MFCC/FBank/谱图。本示例为了课堂快速运行，构造一个二维特征数据集，模拟四个英文数字 `zero/one/two/three` 的声学特征分布。

```python
labels = np.array(['zero', 'one', 'two', 'three'])
centers = np.array([
    [0.0, 0.0],
    [2.5, 0.5],
    [0.5, 2.5],
    [2.6, 2.7],
])

X_parts = []
y_parts = []
for class_id, center in enumerate(centers):
    features = center + np.random.normal(scale=0.55, size=(120, 2))
    X_parts.append(features)
    y_parts.append(np.full(120, class_id))

X = np.vstack(X_parts)
y = np.concatenate(y_parts)
print('样本数:', X.shape[0], '特征维度:', X.shape[1], '类别:', labels.tolist())
```

```python
ask_until_correct(
    title='为什么要划分训练集和测试集',
    question='模型训练后，应该用训练集本身评价最终泛化效果吗？回答：应该 / 不应该',
    answer_set={'不应该', '不', '不能'},
    hint='训练集成绩高，不一定代表新语音也识别得好。',
    knowledge='测试集用于模拟模型没有见过的新语音。如果只看训练集准确率，可能掩盖过拟合问题。语音识别系统最终关心的是对新说话人、新环境、新内容的识别能力。'
)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print('训练集:', X_train_scaled.shape, '测试集:', X_test_scaled.shape)
```

```python
ask_until_correct(
    title='DNN 输出层的含义',
    question='如果要识别 4 个数字类别，DNN 输出层通常应有几个输出单元？回答数字即可',
    answer_set={'4', '四', '四个'},
    hint='每个输出单元可对应一个类别的概率。',
    knowledge='分类任务中，输出层单元数通常等于类别数。语音识别声学模型中，输出可能对应音素、HMM 状态、字符、子词或词片段，取决于建模方案。'
)
```

```python
model = MLPClassifier(
    hidden_layer_sizes=(16, 8),
    activation='relu',
    solver='adam',
    max_iter=800,
    random_state=42
)

model.fit(X_train_scaled, y_train)
pred = model.predict(X_test_scaled)
acc = accuracy_score(y_test, pred)

print('测试集准确率:', round(acc, 4))
print('\n混淆矩阵：')
print(confusion_matrix(y_test, pred))
print('\n分类报告：')
print(classification_report(y_test, pred, target_names=labels))
```

```python
ask_until_correct(
    title='从声学模型到完整语音识别',
    question='DNN 只给出声学类别概率，完整 ASR 通常还需要把概率序列转换成文字。这个过程一般叫：训练 / 解码',
    answer_set={'解码'},
    hint='它发生在模型输出概率之后。',
    knowledge='解码是把声学模型输出的概率序列转换为词、字或子词序列的过程。传统系统可能结合 HMM、发音词典和语言模型；端到端系统则可能使用 CTC、Attention 或 Transducer 等解码方式。'
)
```
