# 老师端：DNN 用于语音识别 Rubric 诊断

本 notebook 读取学生端导出的 `ASR_DNN_checkpoint_records.json`，根据教师端 rubric 自动生成初步诊断和学习反馈报告。

说明：这里的自动诊断是教学辅助，不是高权重考试判分。教师可以根据课堂口头追问修正每个 checkpoint 的判定。

```python
import json
from pathlib import Path
from collections import Counter
from datetime import datetime

RECORD_PATH = Path('ASR_DNN_checkpoint_records.json')
REPORT_PATH = Path('ASR_DNN_teacher_feedback_report.md')
print('读取文件:', RECORD_PATH.resolve())
```

## 1. 教师端 Rubric

| 档位 | 判定依据 |
|---|---|
| 已掌握 | 能说明核心原因，术语基本准确，并能联系刚才的代码或结果 |
| 部分掌握 | 方向正确，但缺少关键机制，或表达不够完整 |
| 需补讲 | 回答暴露明显误解，需要课堂即时补充讲解 |
| 需教师介入 | 多次无法解释，或暴露出前置知识断点 |

```python
data = json.loads(RECORD_PATH.read_text(encoding='utf-8'))
records = data.get('checkpoint_records', [])
print('学生:', data.get('student_name'), data.get('student_id'))
print('任务:', data.get('task'))
print('checkpoint 数量:', len(records))
```

## 2. 关键词辅助诊断

下面的规则用于自动给出初步判断。它不会展示给学生端。教师可以在下一步手动修正。

```python
RUBRIC = {
    'DNN 声学模型的输入': {
        'required': ['特征', '分帧'],
        'help': '补讲短时平稳、分帧、MFCC/FBank/谱图与 DNN 输入之间的关系。'
    },
    '特征维度与类别数': {
        'required': ['输入', '输出'],
        'help': '补讲输入层维度对应特征，输出层维度对应类别数。'
    },
    '训练集、测试集与泛化': {
        'required': ['测试', '训练'],
        'help': '补讲泛化评价，以及标准化器只在训练集 fit 以避免测试集信息泄漏。'
    },
    'DNN 输出层设计': {
        'required': ['类别', '4'],
        'help': '补讲输出单元数与候选类别数的对应关系。'
    },
    'CrossEntropyLoss 与 logits': {
        'required': ['logits'],
        'help': '补讲 PyTorch CrossEntropyLoss 直接接收 logits，内部包含 log-softmax 相关计算。'
    },
    '梯度清零与参数更新': {
        'required': ['梯度', '更新'],
        'help': '补讲 zero_grad 清空梯度、backward 计算梯度、step 更新参数。'
    },
    '评估模式与解码': {
        'required': ['eval', 'no_grad'],
        'help': '补讲 eval/no_grad 的评估作用，以及 argmax 从 logits 得到类别 id。'
    },
    '从分类评价到 ASR 序列评价': {
        'required': ['序列', '错误'],
        'help': '补讲 CER/WER 如何处理替换、插入、删除等序列错误。'
    }
}

def diagnose(record):
    answer = record.get('answer', '')
    title = record.get('title', '')
    rule = RUBRIC.get(title, {'required': [], 'help': '建议教师结合课堂追问进一步判断。'})
    hits = sum(1 for kw in rule['required'] if kw.lower() in answer.lower())
    if not answer.strip():
        status = '需教师介入'
    elif hits >= len(rule['required']) and len(rule['required']) > 0:
        status = '已掌握'
    elif hits > 0:
        status = '部分掌握'
    else:
        status = '需补讲'
    return status, rule['help']

diagnosed = []
for record in records:
    status, suggestion = diagnose(record)
    item = dict(record)
    item['diagnosis'] = status
    item['suggestion'] = suggestion
    diagnosed.append(item)

for item in diagnosed:
    print(item['step'], item['title'], '=>', item['diagnosis'])
```

## 3. 教师手动修正

如需修正自动诊断，可以在下面的 `manual_overrides` 中按 checkpoint 标题填写新档位。可选值：`已掌握`、`部分掌握`、`需补讲`、`需教师介入`。

```python
manual_overrides = {
    # 'CrossEntropyLoss 与 logits': '部分掌握',
}

for item in diagnosed:
    if item['title'] in manual_overrides:
        item['diagnosis'] = manual_overrides[item['title']]

print('修正完成。')
```

```python
def build_report(data, diagnosed):
    counts = Counter(item['diagnosis'] for item in diagnosed)
    lines = []
    lines.append('# 学生学习反馈报告：DNN 用于语音识别')
    lines.append('')
    lines.append('## 1. 任务摘要')
    lines.append(f"- 学生：{data.get('student_name', '')} {data.get('student_id', '')}")
    lines.append(f"- 任务：{data.get('task', 'DNN 用于语音识别')}")
    if data.get('accuracy') is not None:
        lines.append(f"- 测试集准确率：{data['accuracy']:.4f}")
    lines.append(f"- 报告生成时间：{datetime.now().isoformat(timespec='seconds')}")
    lines.append('')
    lines.append('## 2. Checkpoint 统计')
    for status in ['已掌握', '部分掌握', '需补讲', '需教师介入']:
        lines.append(f'- {status}：{counts.get(status, 0)}')
    lines.append('')
    lines.append('## 3. 逐项诊断')
    for i, item in enumerate(diagnosed, start=1):
        lines.append(f"{i}. **{item['title']}**：{item['diagnosis']}")
        lines.append(f"   - 学生回答：{item.get('answer', '')}")
        lines.append(f"   - 自评：{item.get('confidence', '')}")
        lines.append(f"   - 教师建议：{item['suggestion']}")
    lines.append('')
    lines.append('## 4. 后续教学建议')
    weak = [item for item in diagnosed if item['diagnosis'] in {'部分掌握', '需补讲', '需教师介入'}]
    if not weak:
        lines.append('- 本次未发现明显薄弱点，可安排 CTC 或真实音频特征提取的进阶任务。')
    else:
        for item in weak:
            lines.append(f"- 针对“{item['title']}”：{item['suggestion']}")
    return '\n'.join(lines)

report = build_report(data, diagnosed)
REPORT_PATH.write_text(report, encoding='utf-8')
print(report)
print('\n已保存:', REPORT_PATH.resolve())
```
